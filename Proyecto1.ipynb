```python
cursos = []

def agregar_curso():
    nombre = input("Ingrese el nombre del curso: ")
    aula = input("Ingrese el aula del curso: ")
    instructor = input("Ingrese el nombre del instructor: ")
    curso = {
        'nombre': nombre,
        'aula': aula,
        'instructor': instructor,
        'alumnos': []
    }
    cursos.append(curso)
    print(f"Curso '{nombre}' agregado con éxito.")

def eliminar_curso():
    nombre = input("Ingrese el nombre del curso a eliminar: ")
    encontrado = False
    for curso in cursos:
        if curso['nombre'].lower() == nombre.lower():
            cursos.remove(curso)
            print(f"Curso '{nombre}' eliminado con éxito.")
            encontrado = True
            break
    if not encontrado:
        print(f"Curso '{nombre}' no encontrado.")

def modificar_curso():
    nombre = input("Ingrese el nombre del curso a modificar: ")
    encontrado = False
    for curso in cursos:
        if curso['nombre'].lower() == nombre.lower():
            print("Seleccione qué desea modificar:")
            print("1. Nombre del curso")
            print("2. Aula")
            print("3. Instructor")
            print("4. Gestionar alumnos")
            opcion = input("Ingrese el número de la opción: ")

            if opcion == '1':
                nuevo_nombre = input("Ingrese el nuevo nombre del curso: ")
                curso['nombre'] = nuevo_nombre
                print("Nombre del curso modificado.")
            elif opcion == '2':
                nueva_aula = input("Ingrese la nueva aula: ")
                curso['aula'] = nueva_aula
                print("Aula modificada.")
            elif opcion == '3':
                nuevo_instructor = input("Ingrese el nuevo instructor: ")
                curso['instructor'] = nuevo_instructor
                print("Instructor modificado.")
            elif opcion == '4':
                gestionar_alumnos(curso)
            else:
                print("Opción no válida.")
            encontrado = True
            break
    if not encontrado:
        print(f"Curso '{nombre}' no encontrado.")

def gestionar_alumnos(curso):
    while True:
        print(f"\nGestionando alumnos para el curso '{curso['nombre']}'")
        print("1. Agregar alumno")
        print("2. Eliminar alumno")
        print("3. Mostrar alumnos")
        print("4. Volver al menú anterior")
        opcion = input("Ingrese el número de la opción: ")

        if opcion == '1':
            alumno = input("Ingrese el nombre del alumno a agregar: ")
            if alumno not in curso['alumnos']:
                curso['alumnos'].append(alumno)
                print(f"Alumno '{alumno}' agregado al curso '{curso['nombre']}'.")
            else:
                print(f"El alumno '{alumno}' ya está en este curso.")
        elif opcion == '2':
            alumno = input("Ingrese el nombre del alumno a eliminar: ")
            if alumno in curso['alumnos']:
                curso['alumnos'].remove(alumno)
                print(f"Alumno '{alumno}' eliminado del curso '{curso['nombre']}'.")
            else:
                print(f"El alumno '{alumno}' no se encuentra en este curso.")
        elif opcion == '3':
            mostrar_alumnos(curso)
        elif opcion == '4':
            break
        else:
            print("Opción no válida.")


def mostrar_alumnos(curso):
    print(f"\nAlumnos inscritos en '{curso['nombre']}':")
    if curso['alumnos']:
        for alumno in curso['alumnos']:
            print(f"- {alumno}")
    else:
        print("No hay alumnos inscritos en este curso.")

def mostrar_cantidad_alumnos():
    print("\nCantidad de alumnos por curso:")
    if cursos:
        for curso in cursos:
            print(f"- '{curso['nombre']}': {len(curso['alumnos'])} alumnos")
    else:
        print("No hay cursos registrados.")

def mostrar_menu():
    print("\n--- Menú de Gestión de Cursos ISW ---")
    print("1. Agregar curso")
    print("2. Eliminar curso")
    print("3. Modificar curso")
    print("4. Mostrar cantidad de alumnos por curso")
    print("5. Mostrar lista de alumnos de un curso")
    print("6. Salir")

while True:
    mostrar_menu()
    opcion = input("Ingrese el número de la opción: ")

    if opcion == '1':
        agregar_curso()
    elif opcion == '2':
        eliminar_curso()
    elif opcion == '3':
        modificar_curso()
    elif opcion == '4':
        mostrar_cantidad_alumnos()
    elif opcion == '5':
        nombre_curso = input("Ingrese el nombre del curso para mostrar los alumnos: ")
        encontrado = False
        for curso in cursos:
            if curso['nombre'].lower() == nombre_curso.lower():
                mostrar_alumnos(curso)
                encontrado = True
                break
        if not encontrado:
            print(f"Curso '{nombre_curso}' no encontrado.")
    elif opcion == '6':
        print("Saliendo del programa. ¡Hasta luego!")
        break
    else:
        print("Opción no válida. Por favor, intente de nuevo.")
```
