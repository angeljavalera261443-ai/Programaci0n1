## Implement `calcular edad(año nacimiento)` function

### Subtask:
Implement `calcular edad(año nacimiento)` function: This function will take the birth year as input and return the current age.


**Reasoning**:
Implement the `calcular_edad` function as described in the instructions, using the `datetime` module to get the current year.



```python
from datetime import datetime

def calcular_edad(año_nacimiento):
  """Calcula la edad a partir del año de nacimiento."""
  año_actual = datetime.now().year
  edad = año_actual - año_nacimiento
  return edad
```

## Implement `determinar elemento(año nacimiento)` function

### Subtask:
Implement `determinar elemento(año nacimiento)` function: This function will take the birth year as input and return the zodiac element based on the last digit.


**Reasoning**:
Define the `determinar_elemento` function as instructed, calculating the last digit of the birth year and mapping it to a zodiac element using an if-elif-else structure.



```python
def determinar_elemento(año_nacimiento):
  """Determina el elemento del zodiaco basado en el último dígito del año de nacimiento."""
  ultimo_digito = año_nacimiento % 10

  if ultimo_digito == 0 or ultimo_digito == 1:
    return "Metal"
  elif ultimo_digito == 2 or ultimo_digito == 3:
    return "Agua"
  elif ultimo_digito == 4 or ultimo_digito == 5:
    return "Madera"
  elif ultimo_digito == 6 or ultimo_digito == 7:
    return "Fuego"
  elif ultimo_digito == 8 or ultimo_digito == 9:
    return "Tierra"
  else:
      return "Elemento desconocido"
```

## Implement `iniciar oraculo()` function

### Subtask:
Implement `iniciar oraculo()` function: This function will be the main function to control the program flow, get user input, call other functions, and display the final prediction in a formatted way. It should include a loop to allow the user to get multiple predictions.


**Reasoning**:
Implement the `iniciar_oraculo()` function with a loop to get user input, call other functions, and display the results, including input validation for birth year and lucky number.



```python
def iniciar_oraculo():
  while True:
    try:
      nombre = input("Por favor, introduce tu nombre: ")
      año_nacimiento = int(input("Por favor, introduce tu año de nacimiento: "))
      numero_suerte = int(input("Por favor, introduce tu número de la suerte (1-4): "))

      if numero_suerte < 1 or numero_suerte > 4:
        print("Número de la suerte inválido. Debe estar entre 1 y 4.")
        continue

      edad = calcular_edad(año_nacimiento)
      elemento = determinar_elemento(año_nacimiento)
      prediccion = generar_prediccion(nombre, elemento, numero_suerte)

      print("\n--- Tu Predicción del Oráculo ---")
      print(f"Nombre: {nombre}")
      print(f"Edad: {edad}")
      print(f"Elemento del Zodiaco: {elemento}")
      print(f"Tu Número de la Suerte: {numero_suerte}")
      print(f"\n{prediccion}")

    except ValueError:
      print("Entrada inválida. Por favor, introduce un número válido para el año de nacimiento y el número de la suerte.")
      continue

    otra_prediccion = input("¿Quieres otra predicción? (sí/no): ").lower()
    if otra_prediccion != 'sí' and otra_prediccion != 'si':
      break

iniciar_oraculo()
```
