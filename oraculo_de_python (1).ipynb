```python
from datetime import datetime

def calcular_edad(año_nacimiento):
  """Calcula la edad a partir del año de nacimiento."""
  año_actual = datetime.now().year
  edad = año_actual - año_nacimiento
  return edad
```

```python
def determinar_elemento(año_nacimiento):
  """Determina el elemento del zodiaco basado en el último dígito del año de nacimiento."""
  ultimo_digito = año_nacimiento % 10

  if ultimo_digito == 0 or ultimo_digito == 1:
    return "Metal"
  elif ultimo_digito == 2 or ultimo_digito == 3:
    return "Agua"
  elif ultimo_digito == 4 or ultimo_digito == 5:
    return "Madera"
  elif ultimo_digito == 6 or ultimo_digito == 7:
    return "Fuego"
  elif ultimo_digito == 8 or ultimo_digito == 9:
    return "Tierra"
  else:
      return "Elemento desconocido"
```

```python
def generar_prediccion(nombre, elemento, numero_suerte):
  """Genera una predicción personalizada combinando nombre, elemento y número de la suerte."""
  predicciones = {
      1: f"¡Hola, {nombre} del elemento {elemento}! Tu número de la suerte ({numero_suerte}) te trae una semana llena de sorpresas agradables y oportunidades inesperadas. ¡Aprovecha cada momento!",
      2: f"{nombre}, con la energía del elemento {elemento} y la influencia de tu número ({numero_suerte}), encontrarás la fuerza para superar cualquier desafío. Mantén la confianza en ti mismo.",
      3: f"Para ti, {nombre}, regido por el elemento {elemento} y con la suerte del número {numero_suerte}, se abren nuevos caminos en tu vida profesional. Prepárate para el éxito.",
      4: f"Estimado/a {nombre}, tu conexión con el elemento {elemento} y la vibración de tu número ({numero_suerte}) indican que es un momento ideal para fortalecer tus relaciones personales. ¡Disfruta de la compañía de tus seres queridos!"
  }
  return predicciones.get(numero_suerte, f"¡Hola, {nombre} del elemento {elemento}! Tu número de la suerte ({numero_suerte}) te trae una predicción misteriosa. Mantente alerta a las señales del universo.")
```

```python
def iniciar_oraculo():
  while True:
    try:
      nombre = input("Por favor, introduce tu nombre: ")
      año_nacimiento = int(input("Por favor, introduce tu año de nacimiento: "))
      numero_suerte = int(input("Por favor, introduce tu número de la suerte (1-4): "))

      if numero_suerte < 1 or numero_suerte > 4:
        print("Número de la suerte inválido. Debe estar entre 1 y 4.")
        continue

      edad = calcular_edad(año_nacimiento)
      elemento = determinar_elemento(año_nacimiento)
      prediccion = generar_prediccion(nombre, elemento, numero_suerte)

      print("\n--- Tu Predicción del Oráculo ---")
      print(f"Nombre: {nombre}")
      print(f"Edad: {edad}")
      print(f"Elemento del Zodiaco: {elemento}")
      print(f"Tu Número de la Suerte: {numero_suerte}")
      print(f"\n{prediccion}")

    except ValueError:
      print("Entrada inválida. Por favor, introduce un número válido para el año de nacimiento y el número de la suerte.")
      continue

    otra_prediccion = input("¿Quieres otra predicción? (sí/no): ").lower()
    if otra_prediccion != 'sí' and otra_prediccion != 'si':
      break

iniciar_oraculo()
```
